```python
import os

base = "/content/ecosort_yolo_colab"

total_images = 0
total_objects = 0

for split in ["train", "val", "test"]:

    image_dir = f"{base}/images/{split}"
    label_dir = f"{base}/labels/{split}"

    images = [
        f for f in os.listdir(image_dir)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    labels = [
        f for f in os.listdir(label_dir)
        if f.endswith(".txt")
    ]

    objects = 0

    for label in labels:
        with open(os.path.join(label_dir, label), "r") as f:
            objects += sum(1 for line in f if line.strip())

    total_images += len(images)
    total_objects += objects

    print(
        f"{split.upper():5} | "
        f"Images: {len(images):4} | "
        f"Labels: {len(labels):4} | "
        f"Objects: {objects:4}"
    )

print("\nTOTAL IMAGES:", total_images)
print("TOTAL OBJECTS:", total_objects)
```

```python
from ultralytics import YOLO
import torch

torch.manual_seed(42)

# Pretrained lightweight detector
model = YOLO("yolo11n.pt")

results = model.train(
    data="/content/ecosort_yolo_colab/ecosort.yaml",

    # Training
    epochs=80,
    patience=15,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,

    # Reproducibility
    seed=42,
    deterministic=True,

    # Optimizer
    optimizer="AdamW",
    lr0=0.001,
    weight_decay=0.0005,

    # Useful augmentation for real-world waste images
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10.0,
    translate=0.1,
    scale=0.4,
    fliplr=0.5,
    flipud=0.1,
    mosaic=1.0,

    # Disable mosaic near end for cleaner fine-tuning
    close_mosaic=10,

    # Save everything
    save=True,
    plots=True,

    project="/content/ecosort_runs",
    name="baseline_yolo11n_640",
    exist_ok=True,

    verbose=True
)
```

```python
import shutil
import os

run_dir = "/content/ecosort_runs/baseline_yolo11n_640"
output_zip = "/content/baseline_yolo11n_640"

shutil.make_archive(
    output_zip,
    "zip",
    run_dir
)

print("Baseline archive created:")
print(output_zip + ".zip")

print(
    "Size:",
    round(
        os.path.getsize(output_zip + ".zip") / (1024 * 1024),
        2
    ),
    "MB"
)
```
